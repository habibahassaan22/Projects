```python
pop_filepath="/kaggle/input/datasets/rajkumarpandey02/2023-world-population-by-country/countries-table.csv"
pop_data=pd.read_csv(pop_filepath)

first_five_pop_data=pop_data.head(5)
first_five_pop_data
```

```python
#Scatter Plot of First 5 rows

plt.figure(figsize=(6,4))
plt.title("Scatter Plot of First 5 rows to analyze growth Rate")

sns.scatterplot(x=first_five_pop_data["country"],y=first_five_pop_data["growthRate"])
```

```python
#Line Chart for First 5 rows

plt.figure(figsize=(14,10))
plt.title("Line Chart for First 5 rows")
sns.lineplot(data=first_five_pop_data)
```

```python
#Bar Chart
plt.figure(figsize=(10,6))
plt.title("Bar Chart")
sns.barplot(x=first_five_pop_data["country"],y=first_five_pop_data["area"])
```
